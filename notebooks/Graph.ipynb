```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# ✍️ DATA ENTRY SECTION (Edit this!)
# ==========================================
# Replace the numbers below with your actual results.
# You can add or remove models from this list.

data = {
    'Model_Name': [

        'NB_BOW_Upsampled', 
        'NB_TFIDF_Upsampled',
        'linear_SVM_TFIDF_Upsampled',
        'logistic_regression_TFIDF_Upsampled'
        # Add your teammate's models here later if you want!
    ],
    # Enter accuracy percentages (0-100)
    'E/I': [ 71.58, 73.64, 79.03, 77.14], 
    'S/N': [ 78.79, 78.89, 86.83, 86.06],
    'F/T': [ 77.88, 76.67, 78.09, 69.40],
   'P/J': [ 66.65, 67.19, 69.58, 60.84],  
}

# Create the DataFrame
df = pd.DataFrame(data)
df = df.set_index('Model_Name') # Set model names as the index for easier plotting

# ==========================================
# 📊 GRAPH 1: RECREATION OF THE IMAGE
# (Y-axis = Models, Colors = Dimensions)
# ==========================================
# We define the colors to match your uploaded image (Blue, Red, Yellow, Green)
colors = ['#4285F4', '#EA4335', '#FBBC05', '#34A853']

plt.figure(figsize=(10, 6))
ax1 = df.plot(kind='barh', figsize=(12, 7), color=colors, width=0.8)

plt.title('Model Accuracy Comparison (By Model)', fontsize=16, fontweight='bold')
plt.xlabel('Accuracy (%)', fontsize=12)
plt.ylabel('Model', fontsize=12)
plt.xlim(0, 100) # Fix x-axis from 0 to 100
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Dimensions")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```

```python
# ==========================================
# 📊 GRAPH 2: TRANSPOSED COMPARISON
# (Y-axis = Dimensions, Colors = Models)
# ==========================================
# We "transpose" (.T) the data so Dimensions become rows and Models become columns
df_transposed = df.T 

plt.figure(figsize=(10, 6))
# We use a distinct color map ('tab10') so each model gets a clear unique color
ax2 = df_transposed.plot(kind='barh', figsize=(12, 7), colormap='tab10', width=0.8)

plt.title('Best Model per Personality Segment', fontsize=16, fontweight='bold')
plt.xlabel('Accuracy (%)', fontsize=12)
plt.ylabel('Personality Dimension', fontsize=12)
plt.xlim(0, 100)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Models")
plt.grid(axis='x', linestyle='--', alpha=0.7)

# --- THE FIX: Invert the Y-axis so the first item is at the top ---
ax2.invert_yaxis()  

# Add value labels on the bars for easier reading
for container in ax2.containers:
    ax2.bar_label(container, fmt='%.1f%%', padding=3)

plt.tight_layout()
plt.show()
```
